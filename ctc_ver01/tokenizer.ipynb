```python
from datasets import  load_dataset

dataset = load_dataset(
    "openslr/librispeech_asr",
    "clean",
    split="train.100"
)
```

```python
from tokenizer import TokenizerBPE01
```

```python
len(list(dataset['text']))
```

```python
ini_vocab = ["<PAD>"]
tokenizer = TokenizerBPE01(initial_vocab=ini_vocab)
tokenizer2 = TokenizerBPE01(initial_vocab=ini_vocab)
huge_text = ""
for text in list(dataset['text']):
    huge_text += (text + ' ')
```

```python
mincount = 30
mincount2 = 10
tokenizer.train(huge_text, 10000, min_count=mincount)
tokenizer2.train(huge_text, 10000, min_count=mincount2)
```

```python
# Save
save_path = "vocab.json"
save_path2 = "vocab2.json"
tokenizer.saveToJSON(save_path)
tokenizer2.saveToJSON(save_path2)
```

```python
import tkinter as tk
def open_popup():
    popup = tk.Toplevel(root)
    popup.title("Simple Pop-up")
    label = tk.Label(popup, text="This is a pop-up window.")
    label.pack(padx=20, pady=20)
    button = tk.Button(popup, text="Close", command=popup.destroy)
    button.pack(pady=10)
root = tk.Tk()
root.title("Main Window")
main_button = tk.Button(root, text="Finished", command=open_popup)
main_button.pack(pady=20)
root.mainloop()
```
